#### Ollama
Ollama supports embedding models, making it possible to build retrieval augmented generation (RAG) applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
```

```python
embeddings=(
    OllamaEmbeddings(model="gemma:2b")  ##by default it ues llama2
)
```

```python
embeddings
```

```python
r1=embeddings.embed_documents(
    [
       "Alpha is the first letter of Greek alphabet",
       "Beta is the second letter of Greek alphabet", 
        "Teja is the third letter of Greek alphabet",
    ]
)

#in this step it is translation phase where human language is converted into a format that a computer can actually perform math on
''' 
1. Vectorization (The "What")
Computers don't understand the meaning of words like "Alpha" or "Greek." Instead, this function sends your list of strings to a pre-trained model (like OpenAI’s text-embedding-3-small or a HuggingFace model).

The model converts each sentence into a Vector: a long list of floating-point numbers (e.g., [0.12, -0.05, 0.88, ...]).

2. Capturing Semantic Meaning (The "How")
These numbers aren't random. They represent the coordinates of that sentence in a high-dimensional space.

Sentences with similar meanings are placed close together.

Sentences with different meanings are placed far apart.

For example, your first two sentences about "Alpha" and "Beta" will result in vectors that are mathematically very similar because their context (Greek alphabet) is almost identical.
'''
```

```python
len(r1[0]) 

#2048 means that each sentence is represented as a vector of 2048 dimensions. This high-dimensional representation allows the model to capture complex semantic relationships between sentences, enabling it to understand and compare their meanings effectively.
```

```python
embeddings.embed_query("What is the second letter of Greek alphabet ")
```
